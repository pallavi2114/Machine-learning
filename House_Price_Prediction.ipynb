```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
import sklearn.datasets
from sklearn.model_selection import train_test_split
from sklearn import metrics
from xgboost import XGBRegressor
```

```python
dataset = sklearn.datasets.load_boston()
```

```python
dataset = pd.read_csv('/content/HousingData.csv')
dataset.head()
```

```python
dataset.shape

#check for missing or null values
dataset.isnull().sum()
```

```python
#replace NA values with 0
dataset.replace(to_replace=np.nan,value=0)
```

```python
dataset.describe()
```

```python
#lets find the correlation within the features

correlation = dataset.corr()

#data visualization
plt.figure(figsize=(10,10))
sb.heatmap(correlation,cbar=True,square=True,fmt='.1f',annot=True,annot_kws={'size':8},cmap='Blues')
```

```python
#Before we start training, let us rename our label i.e.MEDV
dataset.rename(columns={'MEDV':'PRICE'},inplace=True)
print(dataset)
```

```python
x = dataset.drop(['PRICE'],axis=1)
y = dataset['PRICE']
print(x)
print(y)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
```

```python
model = XGBRegressor()
model.fit(x_train,y_train)
```

```python
train_prediction = model.predict(x_train)
print(train_prediction)
```

```python
#R squared error
score1 = metrics.r2_score(y_train,train_prediction)
#Mean absolute error
score2 = metrics.mean_absolute_error(y_train,train_prediction)
print('R squared error : ',score1)
print('Mean absolute error : ',score2)
```

```python
plt.scatter(y_train,train_prediction)
plt.xlabel('Actual prices')
plt.ylabel('Predicted prices')
plt.title('Actual VS Predicted prices')
plt.show()
```

```python
#Testing the model
test_prediction = model.predict(x_test)

#R squared error
score1 = metrics.r2_score(y_test,test_prediction)
#Mean absolute error
score2 = metrics.mean_absolute_error(y_test,test_prediction)
print('R squared error : ',score1)
print('Mean absolute error : ',score2)
```

```python
plt.scatter(y_test,test_prediction)
plt.xlabel('Actual prices')
plt.ylabel('Predicted prices')
plt.title('Actual VS Predicted prices')
plt.show()
```
